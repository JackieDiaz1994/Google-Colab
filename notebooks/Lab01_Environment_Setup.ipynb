```python
import pandas as pd

data = {
    "Category": ["Technology", "Furniture", "Office Supplies"],
    "Sales": [125000, 95000, 110000]
}

df = pd.DataFrame(data)

df
```

```python
import matplotlib.pyplot as plt

df.plot(
    x="Category",
    y="Sales",
    kind="bar",
    legend=False
)

plt.title("Sample Sales by Category")
plt.ylabel("Sales")
plt.show()
```
